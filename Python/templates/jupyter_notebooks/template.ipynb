# Library Loading

```python
import numpy as np
import pandas as pd
from pathlib import Path
from IPython.display import display
from datetime import datetime
import logging
import colorlog
LOG_COLORS = {
    "DEBUG": "cyan",
    "INFO": "green",
    "WARNING": "yellow",
    "ERROR": "red",
    "CRITICAL": "red,bg_white",
}

logging.basicConfig(
    level=logging.DEBUG,
    format="%(asctime)s - %(name)s - %(funcName)s - %(levelname)s - %(message)s",
    handlers=[
        logging.FileHandler("debug.log", mode="w", encoding="utf-8"),
        logging.StreamHandler(),
    ],
)

# Replace the StreamHandler's formatter with ColoredFormatter
logging.getLogger().handlers[1].setFormatter(
    colorlog.ColoredFormatter(
        "%(asctime)s - %(name)s - %(funcName)s - %(log_color)s%(levelname)s%(reset)s - %(message)s",
        log_colors=LOG_COLORS,
    )
)

# Set the console handler to a higher level to reduce verbosity
logging.getLogger().handlers[1].setLevel(logging.INFO)
logger = logging.getLogger(__name__)
pd.set_option('display.float_format', lambda x: f'{x:,.2f}'.replace(',', ' ').replace('.', ','))
```

# Util functions

```python
def display_adapted(df: pd.DataFrame, nrows: int=None, ncols: int=None) -> None:
    nr = df.shape[0] if nrows is None else nrows
    nc = df.shape[1] if ncols is None else ncols
    with pd.option_context("display.max_rows", nr, "display.max_columns", nc):
        display(df.iloc[:, list(range(nc))].head(nr))
```
